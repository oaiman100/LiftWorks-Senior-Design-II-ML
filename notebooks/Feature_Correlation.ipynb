```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
input_file = "/content/drive/MyDrive/Test_data/PALMO_NACA_4Series_Complete_Dataset.xlsx"

df = pd.read_excel(
    input_file,
    sheet_name="PALMO_4Series_Data"
)


# ============================================================
# 1. INPUT-ONLY CORRELATION MATRIX
# ============================================================

input_data = df[
    [
        "%Camber",
        "%Thickness",
        "Reynolds#",
        "Mach#",
        "Alpha"
    ]
]

input_corr = input_data.corr(method="spearman")

print("Input Feature Correlation Matrix:")
print(input_corr)

plt.figure(figsize=(8, 6))

sns.heatmap(
    input_corr,
    annot=True,
    cmap="Reds",
    linewidths=1,
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Input Feature Correlation Matrix (Spearman)")
plt.tight_layout()

plt.savefig(
    "PALMO_Input_Correlation_Matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 2. INPUT + OUTPUT CORRELATION MATRIX
# ============================================================

all_data = df[
    [
        "%Camber",
        "%Thickness",
        "Reynolds#",
        "Mach#",
        "Alpha",
        "cl",
        "cd",
        "cm"
    ]
]

all_corr = all_data.corr(method="spearman")

print("\nInput and Output Correlation Matrix:")
print(all_corr)

plt.figure(figsize=(11, 9))

sns.heatmap(
    all_corr,
    annot=True,
    cmap="Reds",
    linewidths=1,
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Input and Output Correlation Matrix (Spearman)")
plt.tight_layout()

plt.savefig(
    "PALMO_Input_Output_Correlation_Matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
